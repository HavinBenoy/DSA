Problem :

Longest Common Prefix

Write a function to find the longest common prefix string amongst an array of strings.
If there is no common prefix, return an empty string "".

 

Example 1:

Input: strs = ["flower","flow","flight"]
Output: "fl"




Example 2:

Input: strs = ["dog","racecar","car"]
Output: ""
Explanation: There is no common prefix among the input strings.
 

```python
# My solution :

def check_prefix(arr, endvalue):
    prefix_word = arr[0][0:endvalue+1]
    for i in arr:
        if i[0:endvalue+1] != prefix_word:
            return False
    return True


def longest_common_prefix(arr):

    # Since its a prefix word , it should always start from the zeroth index
    # So our task is to figure out the end index of the common word
    common_prefix_word = ""
    low = 0
    high = float("inf")
    for i in range(len(arr)):
        if len(arr[i])< high:
            high = len(arr[i])
    while (low<=high):
        mid = low + (high-low)//2
        if check_prefix(arr, endvalue = mid):
            common_prefix_word = arr[0][0:mid+1]
            low = mid+1
        else:
            high = mid-1
    return common_prefix_word

arr = ["dog","racecar","car"]
print(f"Longest common prefix word in the array is : '{longest_common_prefix(arr)}'")
```

```python
# Optimal solution : Vertical Scanning Approach

def longest_common_prefix(strs):
    if not strs:
        return ""

    for i in range(len(strs[0])):

        for word in strs[1:]:

            if i >= len(word) or word[i] != strs[0][i]:
                return strs[0][:i]

    return strs[0]

arr = ["flower","flow","flight"]
print(f"Longest common prefix word in the array is : '{longest_common_prefix(arr)}'")
```
