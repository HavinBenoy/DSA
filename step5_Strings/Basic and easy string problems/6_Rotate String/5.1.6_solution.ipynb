# for example try :

# so instead of checking only one case in which an element in goal matches the first element in s , we will check all the possibilities in which an element in goal is equal to the first element in s.



```python
# My solution:

def isRotatedString(s,goal):

    if len(s) != len(goal):
        return False

    start_position_in_goal = -1
    for k in range(len(goal)):
        if goal[k] != s[0]:
            continue
        result = True
        start_position_in_goal = k
        i = 0
        for j in range(start_position_in_goal,len(goal)):
            if s[i] != goal[j]:
                result = False
                break
            i += 1
            
        if result != False and start_position_in_goal != 0 :
            for j in range(start_position_in_goal):
                if goal[j] != s[i]:
                    result =  False
                i += 1
        if result==True:
            return True
    return False

s = "abac"
goal = "acab"
print(isRotatedString(s,goal))
```
