Problem Statement: Given an integer array arr of size N, sorted in ascending order (with distinct values). Now the array is rotated between 1 to N times which is unknown. Find how many times the array has been rotated.



Examples:

Input : arr = [4,5,6,7,0,1,2,3]
Result: 4
Explanation: The original array should be [0,1,2,3,4,5,6,7]. So, we can notice that the array has been rotated 4 times.

Input : arr = [3,4,5,1,2]
Output : 3
Explanation: The original array should be [1,2,3,4,5]. So, we can notice that the array has been rotated 3 times.

```python
# My solution 

def number_of_times_array_sorted(arr):
    low = 0
    high = len(arr)-1
    min_element = float('+inf')
    min_elem_index = -1
    while (low<=high):
        mid = low + (high-low)//2
        if arr[mid]>=arr[low]:
            if min_element>arr[low]:
                min_elem_index = low
                min_element = arr[min_elem_index]
            low = mid+1
        else:
            if min_element>arr[mid]:
                min_elem_index = mid
                min_element = arr[min_elem_index]
            high = mid-1
    return min_elem_index # If the min elem moved to index 1 then array rotated 1 time . min elem in index 2 means it has been roated 2 times. LIkewise to see how many times the array is rotated we simply needs to return the index of the min element

arr = [3,4,5,1,2]

res = number_of_times_array_sorted(arr)

print(f"The array has been rotated {res} times")
            
```
