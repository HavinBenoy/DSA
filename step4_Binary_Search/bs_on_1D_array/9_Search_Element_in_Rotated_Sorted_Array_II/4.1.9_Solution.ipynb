Problem : Given an integer array arr of size N, sorted in ascending order (may contain duplicate values) and a target value k. Now the array is rotated at some pivot point unknown to you. Return True if k is present and otherwise, return False.

```python
#My solution : Better than the bruteforce approach

def find_target(arr,target):
    low = 0
    high = len(arr)-1

    while (low<=high):
        if arr[low]==target:
            return low
        
        while (arr[low]==arr[high] and low!=high):
            low = low+1
        if low==high:
            if arr[low]==target:
                return low
            else:
                return -1
        
        mid = (low+high)//2
        if arr[mid]==target:
            return mid
        
        if (arr[low]==arr[mid-1]):
            if arr[low]==target:
                return low
            else:
                low = mid+1
        elif (arr[low]<arr[mid-1]):
            if arr[low]<=target<=arr[mid-1]:
                high = mid-1
            else:
                low = mid + 1
        else:
            if arr[mid+1]<=target<=arr[high]:
                low = mid+1
            else:
                high = mid -1

    return -1

array = [7, 8, 1, 2, 3, 3, 3, 4, 5, 6]
target = 3

result = find_target(array,target)

if result==-1:
    print(f"{target} is not in the array")
else:
    print(f"The target {target} is found in the index {result}")
        

```

```python
# Solution : Bruteforce Approach

class Solution:
    def search_in_rotated_sorted_array_II(self, arr, k):
        """
        Performs linear search in a rotated sorted array that may contain duplicates.
        
        Args:
            arr (List[int]): The input array
            k (int): The target value to find
        
        Returns:
            bool: True if target is found, else False
        """
        for num in arr:
            if num == k:
                return True  # Target found
        return False  # Target not found


if __name__ == "__main__":
    # Define the rotated sorted array with duplicates
    arr = [7, 8, 1, 2, 3, 3, 3, 4, 5, 6]
    k = 3  # Target value to search

    # Create an instance of Solution
    sol = Solution()

    # Call the search method
    result = sol.search_in_rotated_sorted_array_II(arr, k)

    # Display result
    if result:
        print("Target is present in the array.")
    else:
        print("Target is not present.")
```

```python
# Solution : Optimal Approach

class Solution:
    def search_in_rotated_sorted_array_ii(self, arr, k):
        low, high = 0, len(arr) - 1

        while low <= high:
            mid = (low + high) // 2

            # If mid is the target
            if arr[mid] == k:
                return True

            # Cannot determine sorted half due to duplicates
            if arr[low] == arr[mid] == arr[high]:
                low += 1
                high -= 1
                continue

            # Left half is sorted
            if arr[low] <= arr[mid]:
                if arr[low] <= k <= arr[mid]:
                    high = mid - 1
                else:
                    low = mid + 1
            else:
                # Right half is sorted
                if arr[mid] <= k <= arr[high]:
                    low = mid + 1
                else:
                    high = mid - 1

        return False


if __name__ == "__main__":
    arr = [7, 8, 1, 2, 3, 3, 3, 4, 5, 6]
    k = 3
    sol = Solution()
    result = sol.search_in_rotated_sorted_array_ii(arr, k)

    if result:
        print("Target is present in the array.")
    else:
        print("Target is not present.")
```
