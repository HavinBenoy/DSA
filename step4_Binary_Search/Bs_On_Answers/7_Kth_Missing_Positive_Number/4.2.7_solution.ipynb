Kth Missing Positive Number

Problem Statement: You are given a strictly increasing array ‘vec’ and a positive integer 'k'. Find the 'kth' positive integer missing from 'vec'.

            


Examples


Example 1:
Input Format: vec[]={4,7,9,10}, k = 1
Result: 1
Explanation: The missing numbers are 1, 2, 3, 5, 6, 8, 11, 12, ……, and so on. Since 'k' is 1, the first missing element is 1.


Example 2:
Input Format: vec[]={4,7,9,10}, k = 4
Result: 5
Explanation: The missing numbers are 1, 2, 3, 5, 6, 8, 11, 12, ……, and so on. Since 'k' is 4, the fourth missing element is 5.


```python
# My solution : Bruteforce or optimal approaches are far less complex than this so try them :)

# This solution will check the difference between the elements of the list 'vec'.
# Counting the difference will let us know how many missing values are passed
# count of missing values passed is greater than or equal to k , then we know that the missing value we are looking for is present between the two consecutive 'vec' elements
# if the vec is completely checked and still we didnt got the kth missing element. then it will be 'k+len(vec)'. Think in terms of an example to understand



def kth_missing_value(vec,k):
    current = 0
    next = vec[0]
    total = 0
    n = len(vec)

    for i in range(n):
        prev_total = total
        total += (next-current-1)

        if total>=k:
            for j in range(current+1,next):
                prev_total += 1
                if prev_total == k:
                    return j
        
        current = next
        if i != n-1:
            next = vec[i+1]
    return k+n

vec = [4,7,9,10]
k = 4

result = kth_missing_value(vec,k)
print(result)
```

```python
# Solution : Bruteforce

class MissingKFinder:
    # Function to find the k-th missing number
    def missing_k(self, vec, k):
        for num in vec:
            if num <= k:
                k += 1  # Increase k since num is not missing
            else:
                break  # Stop if num is greater than k
        return k  # Final k is the k-th missing number

# Driver code
vec = [4, 7, 9, 10]
k = 4

finder = MissingKFinder()
ans = finder.missing_k(vec, k)

print("The missing number is:", ans)
```

```python
# Solution : Optimal

class MissingKFinder:
    # Binary search to find the k-th missing number
    def missing_k(self, vec, k):
        low, high = 0, len(vec) - 1

        while low <= high:
            mid = (low + high) // 2

            # Number of missing numbers before index mid
            missing = vec[mid] - (mid + 1)

            if missing < k:
                low = mid + 1  # Need more missing values, go right
            else:
                high = mid - 1  # Too many missing, go left

        # Final k-th missing number calculation
        return k + high + 1     # Personal Note : Here (High+1) means len(vec). The situation in which the program reaches this condition is when the list kth missing value is greater than all the elements in the vec. so once vec is fully checked , we can directly return k+len(vec). 

# Driver code
vec = [4, 7, 9, 10]
k = 4

finder = MissingKFinder()
ans = finder.missing_k(vec, k)

print("The missing number is:", ans)
```
