# Problem :

Nth Root of a Number using Binary Search

Problem Statement: Given two numbers N and M, find the Nth root of M. The nth root of a number M is defined as a number X when raised to the power N equals M. If the 'nth root is not an integer, return -1.



Examples

Input: N = 3, M = 27
Output: 3
Explanation: The cube root of 27 is equal to 3.


Input : N = 4, M = 69
Output: -1
Explanation : The 4th root of 69 does not exist. So, the answer is -1.

```python
# My solution :

def value_of_x(n,m):
    i = 1
    low = 1
    high = m

    while (low<=high):
        mid = (low+high)//2

        power_value_of_current_mid = mid ** n

        if power_value_of_current_mid == m:
            return mid
        elif power_value_of_current_mid<m:
            low = mid+1
        else:
            high = mid-1
    
    return -1


n = 4
m = 39


result  = value_of_x(n,m)

if result !=-1:
    print(f"The {n}th root of {m} = {result}")
else:
    print(f"The {n}th root of {m} doesnt exist")

```

```python
# Solution : Bruteforce

class Solution:
    # Function to find Nth root of M
    def nthRoot(self, n, m):
        # Loop from 1 to m
        for i in range(1, m + 1):
            # Compute i^n
            power = i ** n

            # If equal to m, return i
            if power == m:
                return i

            # If exceeds m, break
            if power > m:
                break

        # If not found, return -1
        return -1

# Main execution
sol = Solution()
n = 3
m = 27

# Find nth root
print("Nth Root:", sol.nthRoot(n, m))
```

```python
# Solution : Optiml apprach

class Solution:
    # Function to find N-th root of M using binary search
    def nthRoot(self, n, m):
        # Set low and high for binary search
        low, high = 1, m

        # Start binary search
        while low <= high:
            # Calculate mid
            mid = (low + high) // 2

            # Store result of mid^n
            ans = 1
            for _ in range(n):
                ans *= mid
                if ans > m:
                    break

            # If mid^n equals m
            if ans == m:
                return mid

            # If mid^n is less than m
            if ans < m:
                low = mid + 1

            # If mid^n is more than m
            else:
                high = mid - 1

        # Return -1 if not found
        return -1

# Driver code
obj = Solution()
result = obj.nthRoot(3, 27)
```
